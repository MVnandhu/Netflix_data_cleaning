### Handle the missing values

```python
df['director']=df['director'].fillna('Unknown')
df['cast']=df['cast'].fillna('Unknown')
df['country']=df['country'].fillna('Unknown')
```

##### The director, cast and country colcumn contain more % of missing value. So, if i drop the rows, nearly  a third of the entire dataset will be gone. Therefore, we are using fillna(), to fill the missing values with 'Unknown'.

```python
df.isnull().sum()
```

```python
df=df.dropna(subset=['date_added','rating','duration'])
```

##### Why drop rows for date_added, rating, duration because the missing values are in less %,these columns are harder to meaingfully fill. Date_added - you can't guess when Netflix added a show, raring - filling with unkown is fine, but since so few rows are affected, dropping them is simple and cost nothing. Duration- Can't guess the runtime.

```python
df.isnull().sum()
```

```python
df.dtypes
```

### Fix the Data type

```python
df['date_added'] = pd.to_datetime(df['date_added'].str.strip())
```

```python
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month
```

```python
df['date_added'].dtype
```

```python
df.dtypes
```

### Remove duplicates


```python
# check the exact duplicate rows 
print(df.duplicated().sum())
```

##### There is no identical rows (ie) every rows have unique value , but it doesn't mean there is no duplicate. let's check  with more targeted subset 

```python
print(df.duplicated(subset=['title','type','release_year']).sum())
```

##### Why title, type, relase_year together, these three columns act like unique identifier for a show 

```python
print(df.shape)
```

```python
df=df.drop_duplicates(subset=['title','type','release_year'])
```

```python
print(df.shape)
```

```python
df['country'].value_counts()
```

##### Standardize text columns

```python
df['country_list']=df['country'].str.split(', ')
```

```python
df['country_list'].head(10)
```

```python
df[['country','country_list']].head(10)
```

```python
df['listed_in'].value_counts()
```

```python
df['genres']=df['listed_in'].str.split(', ')
```

```python
df['genres'].head(10)
```

```python
df['duration_num']=df['duration'].str.extract(r'(\d+)').astype(float)
df['duration_unit']=df['duration'].str.extract(r'([A-Za-z]+)')
```

```python
df[['type', 'duration', 'duration_num', 'duration_unit']].head(10)
```

### Final Validation 

```python
assert(df['release_year']<=2026).all()
assert df['type'].isin(['Movie','TV Show']).all()
print(df[['duration_num','duration_unit']].isnull().sum())
```
